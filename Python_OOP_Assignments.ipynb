```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self._marks = 0
        self.__grade = "F"
        self.set_marks(marks)

    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self._marks = marks
            if marks >= 80:
                self.__grade = "A"
            elif marks >= 60:
                self.__grade = "B"
            elif marks >= 40:
                self.__grade = "C"
            else:
                self.__grade = "F"
            print("Marks updated successfully.")
        else:
            print("Invalid marks. Enter a value from 0 to 100.")

    def get_marks(self):
        return self._marks

    def get_grade(self):
        return self.__grade


student = Student("Bilal", 75)

print("Name:", student.name)
print("Marks:", student.get_marks())
print("Grade:", student.get_grade())

student.set_marks(88)
print("Updated marks:", student.get_marks())
print("Updated grade:", student.get_grade())

student.set_marks(120)

try:
    print(student.__grade)
except AttributeError:
    print("Direct access to __grade is not allowed.")
```

```python
class BankAccount:
    def __init__(self, account_holder, balance, pin):
        self.account_holder = account_holder
        self.__balance = balance
        self.__pin = pin

    def deposit(self, amount):
        if amount <= 0:
            print("Rejected: deposit must be greater than zero.")
        else:
            self.__balance += amount
            print("Deposit successful:", amount)

    def withdraw(self, amount, pin):
        if pin != self.__pin:
            print("Rejected: incorrect PIN.")
        elif amount <= 0:
            print("Rejected: withdrawal must be greater than zero.")
        elif amount > self.__balance:
            print("Rejected: insufficient balance.")
        else:
            self.__balance -= amount
            print("Withdrawal successful:", amount)

    def get_balance(self, pin):
        if pin == self.__pin:
            return self.__balance
        print("Rejected: incorrect PIN.")
        return None


account = BankAccount("Bilal", 10000, 1234)

account.deposit(3000)
account.withdraw(2000, 1234)
account.withdraw(500, 1111)
account.deposit(-500)

print("Current balance:", account.get_balance(1234))
```

```python
class Employee:
    def __init__(self, name, employee_id):
        self.name = name
        self.employee_id = employee_id

    def display_details(self):
        print("Name:", self.name)
        print("Employee ID:", self.employee_id)


class Developer(Employee):
    def __init__(self, name, employee_id, programming_language):
        super().__init__(name, employee_id)
        self.programming_language = programming_language

    def display_details(self):
        super().display_details()
        print("Role: Developer")
        print("Programming Language:", self.programming_language)


class Manager(Employee):
    def __init__(self, name, employee_id, team_size):
        super().__init__(name, employee_id)
        self.team_size = team_size

    def display_details(self):
        super().display_details()
        print("Role: Manager")
        print("Team Size:", self.team_size)


developer = Developer("Ali", 101, "Python")
manager = Manager("Ahmed", 102, 8)

print("Developer Details")
developer.display_details()

print("\nManager Details")
manager.display_details()
```

```python
class Vehicle:
    def __init__(self, brand, model, daily_rental_rate):
        self.brand = brand
        self.model = model
        self.daily_rental_rate = daily_rental_rate

    def calculate_rent(self, days):
        if days <= 0:
            raise ValueError("Rental duration must be greater than zero.")
        return self.daily_rental_rate * days


class Car(Vehicle):
    def calculate_rent(self, days):
        base_rent = super().calculate_rent(days)
        return base_rent + 1000


class Bike(Vehicle):
    def calculate_rent(self, days):
        base_rent = super().calculate_rent(days)
        return base_rent * 0.90


car = Car("Toyota", "Corolla", 5000)
bike = Bike("Honda", "CD 70", 1500)

print("Car rent for 3 days: Rs.", car.calculate_rent(3))
print("Bike rent for 3 days: Rs.", bike.calculate_rent(3))

try:
    car.calculate_rent(0)
except ValueError as error:
    print("Error:", error)
```

```python
class EmailNotification:
    def send(self, message):
        print("Email notification:", message)


class SMSNotification:
    def send(self, message):
        print("SMS notification:", message)


class PushNotification:
    def send(self, message):
        print("Push notification:", message)


class WhatsAppNotification:
    def send(self, message):
        print("WhatsApp notification:", message)


def notify_all(notifications, message):
    for notification in notifications:
        notification.send(message)


notifications = [
    EmailNotification(),
    SMSNotification(),
    PushNotification(),
    WhatsAppNotification()
]

notify_all(notifications, "Your order has been confirmed.")
```

```python
class RegularCustomer:
    def final_price(self, amount):
        if amount < 0:
            raise ValueError("Bill amount cannot be negative.")
        return amount * 0.95


class PremiumCustomer:
    def final_price(self, amount):
        if amount < 0:
            raise ValueError("Bill amount cannot be negative.")
        return amount * 0.90


class VIPCustomer:
    def final_price(self, amount):
        if amount < 0:
            raise ValueError("Bill amount cannot be negative.")
        return amount * 0.80


class GoldCustomer:
    def final_price(self, amount):
        if amount < 0:
            raise ValueError("Bill amount cannot be negative.")
        return amount * 0.85


def print_receipt(customer, amount):
    final_amount = customer.final_price(amount)
    print("Original amount: Rs.", amount)
    print("Final amount: Rs.", final_amount)
    print()


bill_amount = 10000

print("Regular Customer")
print_receipt(RegularCustomer(), bill_amount)

print("Premium Customer")
print_receipt(PremiumCustomer(), bill_amount)

print("VIP Customer")
print_receipt(VIPCustomer(), bill_amount)

print("Gold Customer")
print_receipt(GoldCustomer(), bill_amount)
```

```python
from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius * self.radius

    def perimeter(self):
        return 2 * 3.14159 * self.radius


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)


circle = Circle(5)
rectangle = Rectangle(4, 6)

print("Circle area:", circle.area())
print("Circle perimeter:", circle.perimeter())

print("Rectangle area:", rectangle.area())
print("Rectangle perimeter:", rectangle.perimeter())

try:
    shape = Shape()
except TypeError as error:
    print("Cannot create Shape directly:", error)
```

```python
from abc import ABC, abstractmethod


class PaymentMethod(ABC):
    @abstractmethod
    def pay(self, amount):
        pass

    @abstractmethod
    def refund(self, amount):
        pass


class CardPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            print("Rejected: amount must be positive.")
            return
        print("Card Payment | Payment | Rs.", amount)

    def refund(self, amount):
        if amount <= 0:
            print("Rejected: amount must be positive.")
            return
        print("Card Payment | Refund | Rs.", amount)


class WalletPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            print("Rejected: amount must be positive.")
            return
        print("Wallet Payment | Payment | Rs.", amount)

    def refund(self, amount):
        if amount <= 0:
            print("Rejected: amount must be positive.")
            return
        print("Wallet Payment | Refund | Rs.", amount)


class BankTransferPayment(PaymentMethod):
    def pay(self, amount):
        if amount <= 0:
            print("Rejected: amount must be positive.")
            return
        print("Bank Transfer | Payment | Rs.", amount)

    def refund(self, amount):
        if amount <= 0:
            print("Rejected: amount must be positive.")
            return
        print("Bank Transfer | Refund | Rs.", amount)


def process_transaction(method, amount):
    method.pay(amount)


card = CardPayment()
wallet = WalletPayment()
bank = BankTransferPayment()

process_transaction(card, 5000)
process_transaction(wallet, 2500)
process_transaction(bank, 7000)

card.refund(1000)
wallet.pay(-500)
```
